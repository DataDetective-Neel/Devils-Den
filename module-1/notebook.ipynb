## 1. Echo Environment (MCP-based)

The simplest environment - it just echoes back what you send it.

```python
# Clone OpenEnv repo to get typed environment clients
import os
import sys

if not os.path.exists('OpenEnv'):
    !git clone https://github.com/meta-pytorch/OpenEnv.git

repo = os.path.abspath('OpenEnv')
sys.path.insert(0, repo)
sys.path.insert(0, os.path.join(repo, 'src'))
```

```python
from envs.echo_env import EchoEnv

# Connect to the hosted Echo environment
with EchoEnv(base_url="https://openenv-echo-env.hf.space").sync() as env:
    # List available tools
    tools = env.list_tools()
    print("Available tools:")
    for tool in tools:
        print(f"  - {tool['name']}: {tool.get('description', 'No description')}")
    
    # Call the echo_message tool
    response = env.call_tool("echo_message", message="Hello, OpenEnv!")
    print(f"\nEcho response: {response}")
```

## 2. OpenSpiel Catch (Standard Interface)

A simple game where you control a paddle to catch falling balls.

```python
from envs.openspiel_env import OpenSpielEnv
from envs.openspiel_env.models import OpenSpielAction

# Connect to the Catch environment
with OpenSpielEnv(base_url="https://openenv-openspiel-catch.hf.space").sync() as env:
    # Reset the environment
    result = env.reset()
    print("Initial observation:")
    print(f"  Legal actions: {result.observation.legal_actions}")
    print(f"  Game phase: {result.observation.game_phase}")
    print(f"  Done: {result.observation.done}")
    
    # Take a step (action 1 = STAY)
    result = env.step(OpenSpielAction(action_id=1, game_name="catch"))
    print(f"\nAfter step:")
    print(f"  Reward: {result.observation.reward}")
    print(f"  Done: {result.observation.done}")
    print(f"  Legal actions: {result.observation.legal_actions}")
```

## 3. TextArena Wordle (Standard Interface)

The popular word-guessing game implemented as an RL environment.

```python
from envs.textarena_env import TextArenaEnv
from envs.textarena_env.models import TextArenaAction

# Connect to the Wordle environment
with TextArenaEnv(base_url="https://openenv-textarena-wordle.hf.space").sync() as env:
    # Reset for a new game
    result = env.reset()
    print("Wordle game started!")
    print(f"Initial prompt: {result.observation.rendered}")
    
    # Make a guess
    guess = "CRANE"
    result = env.step(TextArenaAction(content=guess, game_name="wordle"))
    print(f"\nAfter guessing '{guess}':")
    print(f"Response: {result.observation.rendered}")
    print(f"Done: {result.observation.done}")
    print(f"Reward: {result.observation.reward}")
```

## Key Observations

Notice the pattern:

1. **Same connection pattern**: `with XxxEnv(base_url="...").sync() as env:`
2. **Same methods**: All standard environments use `reset()` and `step(action)`
3. **Type-safe actions**: Each environment defines its own `Action` type
4. **Type-safe observations**: Results have well-defined fields you can autocomplete

This consistency makes it easy to:
- Switch between environments
- Write reusable training code
- Debug issues quickly
